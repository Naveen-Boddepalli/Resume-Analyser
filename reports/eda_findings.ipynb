```python
corr=df.corr(numeric_only=True)
```

```python
corr_salary = corr["salary_package_lpa"].sort_values(ascending=False)

plt.figure(figsize=(8, 6))
corr_salary.plot(kind="bar")
plt.title("Correlation with Salary Package")
plt.show()
```
